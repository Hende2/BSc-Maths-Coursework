<u>**A1**</u>

**We show that $p_1(x) = 2x$ for $p_n(x) := \frac{1}{\sin\theta}\sin((n+1)\theta),\quad \theta := \cos^{-1}(x),\quad x\in[-1,1].$**

*Proof:*
\begin{align*} 
p_1(x) & = \frac{\sin{((1 + 1)\theta)}}{\sin{(\theta)}} \\[5pt]
& = \frac{\sin{(2\cos^{-1}(x))}}{\sin{(\cos^{-1}(x))}} \\[5pt]
& = \frac{2\sin{(\cos^{-1}(x))\cos{(\cos^{-1}(x)})}}{\sin{(\cos^{-1}(x))}} \\[3pt]
& = 2x 
\end{align*}

---

**Now we set out to prove that: $p_n(x) = 2x p_{n-1}(x) - p_{n-2}(x),\quad n=2,3,\ldots$**

*Proof:*
\begin{align*}
p_n(x) & = \frac{\sin{((n+1)\theta)}}{\sin{(\theta)}} \\[4pt]
& = \frac{\sin{(n\theta)}\cos{(\theta)} + \sin{(\theta)}\cos{(n\theta)}}{\sin{(\theta)}} \\[5pt]
& = \frac{2\cos{(\theta)}\sin{(n\theta)} - \cos{(\theta)}\sin{(n\theta)}+ \sin{(\theta)}\cos{(n\theta)}}{\sin{(\theta)}} \\[5pt]
& = 2x\frac{\sin{(n\theta)}}{\sin{(\theta)}}- \frac{\sin{((n-1)\theta)}}{\sin{(\theta)}} \\[4pt]
& = 2xp_{n-1}(x) - p_{n-2}(x) 
\end{align*}

---

**Now by using induction we show $p_n(x)$ is a polynomial of degree $n$:**

**Base Cases**:
\begin{align*}
p_0(x) & = 1 = x^0 \\
p_1(x) & = 2x \\
\end{align*}
Both are polynomials, $p_0(x)$ of degree 0 and $p_1(x)$ of degree 1.

**Inductive hypothesis:**

We assume that $p_k(x)$ is a polynomial of degree k with $k \geq 1$:

**Inductive Step:**
\begin{align*}
p_{k+1}(x) = \underbrace{2xp_k(x)}_{\text{Polynomial of degree k + 1}} - \underbrace{p_{k-1}(x)}_{\text{Polynomial of degree k - 1}}\\ 
\end{align*}
This uses the recurrence relation we proved above which implies $p_{k+1}(x)$ is a polynomial of degree k + 1.

When assumed for $p_k(x)$, identity proven for $p_{k+1}(x)$ and hence by induction the result holds $\forall n \geq 1$

---

**Lastly we show that $\{p_0,\cdots,p_n\}$ forms a basis for $\mathcal{P_n}$ by proving its linear independence:**

$$ \{p_0,\cdots,p_n\} \text{ linearly independent} \Leftrightarrow \sum_{k=0}^{n} a_k p_k = 0 \Rightarrow a_0 = a_1 = \cdots = a_n = 0$$

As the sum of polynomials is a polynomial, $ \: \sum_{k=0}^{n} a_k p_k $ is a polynomial.

**Now assume** $a_n \neq 0, \sum_{k=0}^{n} a_k p_k = 0$, then:

$$\sum_{k=0}^{n} a_k p_k = a_0p_0 + \cdots + a_np_n$$ 

contains $a_np_n$, a non-zero degree n polynomial which contradicts the assumption that the sum is 0. Therefore $a_n = 0$.

**By induction** this argument proves $a_k = 0$ for all $k \in \{0,\ldots,n\} \Leftrightarrow \{p_0,\ldots,p_n\} \text{ are linearly independent.}$

Since 
$$\dim(\mathcal{P_n}) = n+1$$

and the list $\{p_0,\ldots,p_n\}$ of n + 1 polynomials is linearly independent, $\{p_0,\ldots,p_n\}$ spans $\mathcal{P_n}$ and therefore is a basis for $\mathcal{P_n}.$

**We show that the matrix $J_n$ is non-singular.**

*Proof:*

As $J_n \in \mathbb{R}^{n \times n}$ or in other words is a square matrix:
\begin{align*}
J_n \text{ non-singular } 
& \Leftrightarrow J_n \text{ has full row rank} \\[3pt]
& \Leftrightarrow \text{Rows of } J_n \text{ are linearly independent} \\[3pt]
\end{align*}

Let $\mathbf{r}_i$ denote the $i$-th row of $J_n$. Then:

$$
J_n \text{ non-singular } \Leftrightarrow \left(\sum_{i=1}^n b_i\mathbf{r}_i = \mathbf{0} \in \mathbb{R}^{1 \times n} \Rightarrow b_i = 0, \text{ for } i = 1,\ldots,n\right)\\[3pt]
$$

**Assume**:

\begin{align*}
\sum_{i=1}^n b_i\mathbf{r}_i & = \begin{pmatrix}
\sum_{i=1}^{n} b_ip_{i-1}(x_1) & \sum_{i=1}^{n} b_ip_{i-1}(x_2) & \cdots & \sum_{i=1}^{n} b_ip_{i-1}(x_n)
\end{pmatrix} = \begin{pmatrix} 0 & 0 & \cdots & 0 \end{pmatrix}
\end{align*}

This implies: 
\begin{align*}
& \sum_{i=1}^n b_ip_{i-1}(x_j) = 0, \quad j = 1,...,n. \\[4pt]
& \Rightarrow \: b_1 = b_2 = \cdots = b_n = 0  \: \left[\text{Linear independence of $\{p_0,\ldots,p_{n-1}\}$ proven in A1}\right]
\end{align*}

Hence the rows of $J_n$ are linearly independent and $J_n$ is non-singular.

---

**Now we consider the importance of $J_n$ being non-singular:**

As $J_n$ non-singular, $J_n$ has an inverse denoted by $J_n^{-1}$. Note that:

\begin{align*}
J_n\mathbf{w} & = \begin{pmatrix} 
p_0(x_1)w_1 + p_0(x_2)w_2 + \cdots + p_0(x_n)w_n \\ 
p_1(x_1)w_1 + p_1(x_2)w_2 + \cdots + p_1(x_n)w_n \\ 
\vdots \\
p_{n-1}(x_1)w_1 + p_{n-1}(x_2)w_2 + \cdots + p_{n-1}(x_n)w_n
\end{pmatrix} \\[5pt] 
& = \begin{pmatrix}
\sum_{i=1}^n p_0(x_i)w_i \\
\sum_{i=1}^n p_1(x_i)w_i \\
\vdots \\
\sum_{i=1}^n p_{n-1}(x_i)w_i \\
\end{pmatrix} \\[5pt]
\end{align*}

If $\mathbf{w}$ is such that:
$$
\sum_{i=1}^n p_k(x_i)w_i = \begin{cases}
A, & k = 0\\ 0, & k = 1, \ldots, n-1 
\end{cases} \quad \text{where } A := \int_{-1}^1 \sqrt{1-x^2}\, dx.
$$

Then it is also that $\mathbf{w}$ is such that:
$$
J_n\mathbf{w} = \mathbf{b} \quad \text{Where } \mathbf{b} = \begin{pmatrix}A \\ 0 \\ \vdots \\ 0 
\end{pmatrix} \in \mathbb{R}^n $$

As $J_n$ is invertible:
\begin{align*}
J_n^{-1}J_n\mathbf{w} & = J_n^{-1}\mathbf{b} \\
\Rightarrow \mathbf{w} & = J_n^{-1}\mathbf{b}
\end{align*}

Thus, we have found $w_1,\ldots,w_n$ satisfying the necessary condition.

**We show that $\int_{-1}^1 p_n(x)p_m(x)\sqrt{1-x^2}\,dx = 0,\quad n\neq m.$**

*Proof:*

Firstly, consider integrating with respect to $d\theta$ instead of $dx$.

\begin{align*}
\theta & = \cos^{-1}{(x)} \\
x  & = \cos{(\theta)} \\
dx & = -\sin{(\theta)}\, d\theta
\end{align*}

We adjust the limits: 
\begin{align*}
x = 1 & \Rightarrow \cos{(\theta)} = 1 \Rightarrow \theta = 0 \\
x = -1 & \Rightarrow \cos{(\theta)} = -1 \Rightarrow \theta = \pi
\end{align*}

Also, we note that: 
$$
\sqrt{1-x^2} = \sqrt{1-\cos^2{(\theta)}} = \sqrt{\sin^2{(\theta)}} = \sin{(\theta)}
$$

$$
\sin(x)\sin(y) = \frac{1}{2}\left[\cos{(x-y)} - \cos{(x+y)}\right]
$$

Then using our identities we consider the integral:

\begin{align*}
\int_{-1}^1 p_n(x)p_m(x)\sqrt{1-x^2}\,dx & = \int_{\pi}^0 \frac{\sin{((n+1)\theta)}}{\sin{(\theta)}}\frac{\sin{((m+1)\theta)}}{\sin{(\theta)}}\sin{(\theta)}(-\sin{(\theta)})\,d\theta \\[5pt]
& = \int_0^{\pi} \sin{((n+1)\theta)}\sin{((m+1)\theta)}\,d\theta \\[5pt]
& = \frac{1}{2} \left[\int_0^{\pi} \cos{((n-m)\theta)}\,d\theta - \int_0^{\pi} \cos{((n+m+2)\theta)}\,d\theta\right] \\[5pt]
& = \frac{1}{2} \left[ \left[\frac{\sin{((n-m)\theta)}}{n-m}\right]_0^{\pi} - \left[\frac{\sin{((n+m-2)\theta)}}{n+m-2}\right]_0^{\pi} \right] \\[5pt]
\end{align*}
As,
$$
\left[\frac{\sin{(k\theta)}}{k}\right]_0^{\pi} = 0 \text{ for any non-zero integer } k
$$
If $n \neq m$:
$$
\int_{-1}^1 p_n(x)p_m(x)\sqrt{1-x^2}\,dx = \frac{1}{2}\left(0\right) = 0
$$

</u>***A4***</u>

**We show that for any $p(x) \in \mathcal{P}_n$:**
$$
\int_{-1}^1 p(x)\sqrt{1-x^2}\, dx = \sum_{i=1}^n p(x_i)w_i
$$

*Proof:*

As $\{p_0,...,p_n\}$ forms a basis for $\mathcal{P_n}$, any $p(x) \in \mathcal{P_n}$ can be expressed as:

$$
p(x) = \sum_{i=0}^n a_ip_i(x), \quad \text{for some } a_0,\ldots,a_n \in \mathbb{R}
$$

Then:
\begin{align*}
\int_{-1}^1 p(x)\sqrt{1-x^2}\, dx & = \int_{-1}^1 \left(\sum_{i=0}^n a_ip_i(x)\right)\sqrt{1-x^2}\,dx \\[4pt]
& = \sum_{i=0}^n a_i \left(\int_{-1}^1 p_i(x)\sqrt{1-x^2}\,dx\right)\\[4pt]
& = \sum_{i=0}^n a_i \left(\int_{-1}^1 p_i(x)\underbrace{p_0(x)}_{=1}\sqrt{1-x^2}\,dx\right)\\[5pt]
& = \sum_{i=0}^n a_i  \begin{cases}
0 & \text{if } i = 1,...,n  \quad \left[\text{By A3}\right]\\
\int_{-1}^1 \sqrt{1-x^2}\,dx = A & \text{if } i = 0
\end{cases}
\end{align*}

---

**Now we need to extend a statement from A2 to suit our needs.** 

In A2 we showed:
$$
\sum_{i=1}^n p_k(x_i)w_i = 
\begin{cases}
A, & k = 0 \\
0, & k = 1, \dots, n - 1
\end{cases}
\quad \text{where } A := \int_{-1}^1 \sqrt{1 - x^2}\, dx
$$

Now we extend it to cover the case of $k = n$:

As $x_1,\ldots,x_n$ are the roots of $p_n(x)$, we have that:
$$
\sum_{i=1}^n p_n(x_i)w_i = \sum_{i=1}^n 0 \cdot w_i = 0 
$$

Which gives us:

$$
\sum_{i=1}^n p_k(x_i)w_i = 
\begin{cases}
A, & k = 0 \\
0, & k = 1, \dots, n
\end{cases}
$$

---

**Going back to our integral we have:**
\begin{align*}
\int_{-1}^1 p(x)\sqrt{1-x^2}\, dx & = \sum_{i=0}^n a_i \left(\sum_{j=1}^n p_i(x_j)w_j\right) \\[3pt]
& = \sum_{j=1}^n w_j \left( \sum_{i=0}^n a_i p_i(x_j) \right) \\[3pt]
& = \sum_{j=1}^n p(x_j)w_j \\[3pt]
& = \sum_{i=1}^n p(x_i)w_i \\
\end{align*}

This confirms our equation for all $p(x) \in \mathcal{P}_n$.

---

**Now we want to extend the result to $p(x) \in \mathcal{P}_{2n-1}$.** 

*Proof:*

A result from Algebra 1A states that:

$$
\text{For any given } p(x) \in \mathcal{P}_{2n-1} \text{ and } p_n(x) \in \mathcal{P}_n, \, \exists q(x), r(x) \in \mathcal{P}_n \text{ such that:}
$$
$$
p(x) = q(x)p_n(x) + r(x) \quad \text{with } \deg(q) \leq n-1, \, \deg(r) \leq n-1
$$

Then:
\begin{align*}
\int_{-1}^1 p(x)\sqrt{1-x^2}\,dx & = \int_{-1}^1 \left[q(x)p_n(x) + r(x)\right]\sqrt{1-x^2}\,dx \\[4pt]
& = \int_{-1}^1 q(x)p_n(x)\sqrt{1-x^2}\,dx + \underbrace{\int_{-1}^1 r(x)\sqrt{1-x^2}\,dx}_{\text{Exact as } \deg(r) \leq n-1} \\[5pt]
\end{align*}

Narrow our attention to $\int_{-1}^1 q(x)p_n(x)\sqrt{1-x^2}\,dx.$ As $q(x) \in \mathcal{P}_{n-1}$:

$$
q(x) = \sum_{i=0}^{n-1} b_i p_i(x) \quad \text{for some } b_0,\ldots,b_{n-1} \text{ not all zero}
$$

Then:
\begin{align*}
\int_{-1}^1 q(x)p_n(x)\sqrt{1-x^2}\,dx & = \int_{-1}^1 \left(\sum_{i=0}^{n-1} b_ip_i(x)\right)p_n(x)\sqrt{1-x^2}\,dx \\[4pt]
& = \sum_{i=0}^{n-1}b_i\left(\int_{-1}^1 p_i(x)p_n(x)\sqrt{1-x^2}\,dx\right) \\[4pt]
& = 0 \quad \text{[By A3]}
\end{align*}

So:
$$
\int_{-1}^1 p(x)\sqrt{1 - x^2}\, dx = \int_{-1}^1 r(x)\sqrt{1 - x^2}\, dx = \sum_{i=1}^n r(x_i)w_i = \sum_{i=1}^n p(x_i)w_i \: \text{ for all } p(x) \in \mathcal{P}_{2n-1}.
$$

\mathbf{w} & = A \begin{pmatrix}
\frac{1}{\left\| \mathbf{v}^1 \right\|^2} & 0 & \cdots & 0 \\
0 & \frac{1}{\left\| \mathbf{v}^2 \right\|^2} & \ddots & \vdots \\
\vdots & \ddots & \ddots & 0 \\
0 & \cdots & 0 & \frac{1}{\left\| \mathbf{v}^n \right\|^2}
\end{pmatrix}\begin{pmatrix}
p_0(x_1) \\
p_0(x_2) \\
\vdots \\ 
\vdots \\
p_0(x_n) \end{pmatrix} \\[5pt]
& = A \begin{pmatrix}
\frac{1}{\left\| \mathbf{v}^1 \right\|^2} & 0 & \cdots & 0 \\
0 & \frac{1}{\left\| \mathbf{v}^2 \right\|^2} & \ddots & \vdots \\
\vdots & \ddots & \ddots & 0 \\
0 & \cdots & 0 & \frac{1}{\left\| \mathbf{v}^n \right\|^2}
\end{pmatrix} \begin{pmatrix}
v^1_1 \\
v^2_1 \\
\vdots \\ 
\vdots \\
v^n_1 \end{pmatrix} \\[5pt]
& = A \begin{pmatrix}
\frac{v^1_1}{\left\| \mathbf{v}^1 \right\|^2} \\
\frac{v^2_1}{\left\| \mathbf{v}^2 \right\|^2} \\
\vdots \\[5pt]
\frac{v^n_1}{\left\| \mathbf{v}^n \right\|^2}\end{pmatrix} \\
& \Rightarrow w_i = A \cdot \frac{v^i_1}{\left\| \mathbf{v}^i \right\|^2} \quad \text{for } i = 1,2,...,n\\
& \Rightarrow w_i = A \cdot \frac{\left(v^i_1\right)^2}{\left\| \mathbf{v}^i \right\|^2} \quad \left[\text{As } v^i_1 = p_0(x_i) = 1 = \left(p_0(x_i)\right)^2 = \left(v^i_1\right)^2 \right]
\end{align*}

<u>***B1***</u>

```python
import numpy as np
def getquad(n):
    '''
    Returns a vector of quadrature nodes x and weights w of dimension n 
    Input: Integer n: Possible number of points
    Outputs: Vector x: Quadrature nodes x1,...,xn. Vector w: Weights w1,...,wn
    '''
    # First we need to define An
    An = np.zeros((n,n))
    An += np.diag(1/2 * np.ones(n-1),k=1)
    An += np.diag(1/2 * np.ones(n-1),k=-1)
    x, eigenvectors = np.linalg.eig(An)
    wlist = []
    for i in range(n):
        vi = eigenvectors[:, i]
        wi = np.pi / 2 * (vi[0] ** 2) / (np.linalg.norm(vi) ** 2)
        wlist.append(wi)
    w = np.array(wlist)
    return x, w
```

<u>***B2***</u>

```python
import scipy
import matplotlib.pyplot as plt
def myquad(g, x, w):
    '''
    Evaluate sum_{i=1,...,n} w_i g(x_i)
    Inputs: A function g, vector x: quadrature nodes, vector w: quadrature weights
    Output: float: evaluation of the sum above
    '''
    assert len(x) == len(w), "x and w should have the same length"
    result = np.sum(w * g(x))
    return result


# Now let's verify the degree of precision is 2n-1 for n=10, we should have that the 
# error is 0 for x^r with r = 0,1,...,19 and then a nonzero error at r = 20.
n = 10
x, w = getquad(n)
exact_vals = []
quad_vals = []
print('r\t Approximation\t Exact\t     Error\t')
for i in range(2*n + 1):
    func = lambda x: x**i
    quad_val = myquad(func,x,w)
    exact, _ = scipy.integrate.quad(lambda x: x**i * np.sqrt(1 - x**2),-1,1)
    error = abs(exact - quad_val)
    quad_vals.append(quad_val)
    exact_vals.append(exact)
    print('{:d} \t {:.9f} \t {:.9f} {:.9f} '.format(i,quad_val,exact,error))
    
errors = [abs(approx - exact) for approx, exact in zip(quad_vals, exact_vals)]
```

<u>***B3***</u>

```python
# Define our functions.
g1 = lambda x: np.exp(x) 
g2 = lambda x: x * np.sin(x)
g3 = lambda x: np.sqrt(1 - x**2)

# Calculate the exact values.
g1_exact, _ = scipy.integrate.quad(lambda x: np.exp(x) * np.sqrt(1 - x**2),-1,1)
g2_exact, _ = scipy.integrate.quad(lambda x: x * np.sin(x) * np.sqrt(1 - x**2),-1,1)
g3_exact, _ = scipy.integrate.quad(lambda x: np.sqrt(1 - x**2) * np.sqrt(1 - x**2),-1,1)

# Collect the quadrature errors.
errors_g1 = []
errors_g2 = []
errors_g3 = []
for i in range(1,21):
    x, w = getquad(i)
    g1_val = myquad(g1,x,w)
    errors_g1.append(abs(g1_val - g1_exact))
    g2_val = myquad(g2,x,w)
    errors_g2.append(abs(g2_val - g2_exact))
    g3_val = myquad(g3,x,w)
    errors_g3.append(abs(g3_val - g3_exact))

# Now let's generate our plots.
plt.figure(figsize = (10,6))
plt.semilogy(np.arange(1,21),errors_g1,label="Error for $g_1(x) = \exp{(x)}$")
plt.semilogy(np.arange(1,21),errors_g2,label="Error for $g_2(x) = x \sin{(x)}$")
plt.semilogy(np.arange(1,21),errors_g3,label="Error for $g_3(x) = \sqrt{1-x^2}$")
plt.xticks(np.arange(1,21, 1))
plt.xlabel('n')
plt.ylabel('Absolute error')
plt.title('Plots of quadrature error for different functions')
plt.legend()
plt.show()
```

<u>***B4***</u>

The approximation for $I_3 := \int_{-1}^{1} \sqrt{1-x^2}\sqrt{1-x^2}\,dx$ is the slowest to converge due to the properties of its derivative. Let $g_3(x) = \sqrt{1-x^2}$.
$$
\frac{dg_3}{dx} = \frac{d}{dx}\left(\sqrt{1-x^2}\right) = -\frac{x}{\sqrt{1-x^2}}
$$
Note that this derivative is undefined at $x = -1$ and $x = 1$, with gradients tending to $\infty$ and $-\infty$ respectively, and so at the endpoints of the integral our function is not smooth. As a result the integral is harder to approximate closer to the boundary than $g_1(x)$  and $g_2(x)$ which leads to the slower convergence rate.

For our sequence acceleration method, we will choose Aitken's delta-squared process. This is because the error for $I_3$ is monotonically decreasing and the convergence appears to be linear from the plot, both good indicators that Aitken's method will accelerate the rate of convergence.

```python
def aitken_acceleration(Q_n):
    '''
    Using aitken's acceleration technique to compute a new sequence that exhibits
    faster convergence.
    Inputs: A list Q_n of length n: The previous sequence for approximating the 
    integral.
    Output: A list A_n of length n - 2: A new sequence for approximating the
    integral.
    '''
    A_n = []
    for i in range(len(Q_n) - 2):
        delta_xn = Q_n[i+1] - Q_n[i]
        delta_xnplusone = Q_n[i+2] - Q_n[i+1]
        delta_xn_sq = delta_xnplusone - delta_xn
        A_n.append(Q_n[i] - ((delta_xn ** 2)/ delta_xn_sq))
    return A_n
```

```python
# Let's collect our exact values and our approximations.
errors_g3 = []
g3_vals = []
g3_exact, _ = scipy.integrate.quad(lambda x: np.sqrt(1 - x**2) * np.sqrt(1 - x**2),-1,1)
for i in range(1,21):
    x, w = getquad(i)
    g3_val = myquad(g3,x,w)
    g3_vals.append(g3_val)
    errors_g3.append(abs(g3_val - g3_exact))

# Let's generate our accelerated sequence.
aitken_sequence = aitken_acceleration(g3_vals)
aitken_errors = [abs(approx - g3_exact) for approx in aitken_sequence]

# Now for the plots.
plt.figure(figsize = (10,6)) # Makes the plot bigger.
plt.semilogy(np.arange(1,21),errors_g3,label="Error of original sequence for $g_3(x) = \sqrt{1-x^2}$")
plt.semilogy(np.arange(1,19),aitken_errors,label="Error for Aitken's acceleration sequence for $g_3(x)$") 
# Note that the accelerated sequence is defined for less n than the original.
plt.xticks(np.arange(1,21, 1))
plt.xlabel('n')
plt.ylabel('Absolute Error')
plt.title("Analysis of the effect of Aitken's sequence acceleration")
plt.legend()
plt.show()
```
